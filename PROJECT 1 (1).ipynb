```python
df["date"]=pd.to_datetime(df["date"])
df.dtypes
```

```python
text_columns=["symbol","asset_name","asset_type","region"]
for col in text_columns:
    df[col]=df[col].astype(str).str.strip()
print("text columns cleaned successfully")
```

```python
print("Negative Open values:", (df["open"] < 0).sum())
print("Negative High values:", (df["high"] < 0).sum())
print("Negative Low values:", (df["low"] < 0).sum())
print("Negative Close values:", (df["close"] < 0).sum())
print("Negative Volume values:", (df["volume"] < 0).sum())
```

```python
total_volume = df["volume"].sum()
average_close = df["close"].mean()
highest_close = df["close"].max()
lowest_close = df["close"].min()

print("Total Trading Volume:", total_volume)
print("Average Closing Price:", average_close)
print("Highest Closing Price:", highest_close)
print("Lowest Closing Price:", lowest_close)
```

```python
total_records=len(df)
print("total number of records:",total_records)
```

```python
start_date = df["date"].min()
end_date = df["date"].max()

print("Starting Date:", start_date)
print("Ending Date:", end_date)
```

```python
df["year"]=df["date"].dt.year
df["month"]=df["date"].dt.month
df["year_month"]=df["date"].dt.to_period("M")
print("\ndata cleaning completed successfully")
```

```python
asset_type_summary = df.groupby("asset_type").agg(
    Number_of_Records=("close", "count"),
    Average_Close=("close", "mean"),
    Highest_Close=("close", "max"),
    Total_Volume=("volume", "sum"),
    Average_Volume=("volume", "mean")
).sort_values("Average_Close", ascending=False)

asset_type_summary
```

```python
asset_name_summary = df.groupby("asset_name").agg(
    Number_of_Records=("close", "count"),
    Average_Close=("close", "mean"),
    Highest_Close=("close", "max"),
    Lowest_Close=("close", "min"),
    Total_Volume=("volume", "sum")
).sort_values("Total_Volume", ascending=False)

asset_name_summary
```

```python
region_summary = df.groupby("region").agg(
    Number_of_Records=("close", "count"),
    Average_Close=("close", "mean"),
    Highest_Close=("close", "max"),
    Total_Volume=("volume", "sum")
).sort_values("Total_Volume", ascending=False)
region_summary
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(
    x="asset_type",
    y="close",
    data=df
)
plt.title("Average Closing Price by Asset Type")
plt.xlabel("Asset Type")
plt.ylabel("Average Closing Price")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
asset_type_summary["Total_Volume"].plot(
    kind="bar"
)
plt.title("Total Trading Volume by Asset Type")
plt.xlabel("Asset Type")
plt.ylabel("Total Trading Volume")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df["Year"] = df["date"].dt.year
year_summary = df.groupby("Year")["close"].mean()
year_summary
```

```python
plt.figure(figsize=(12, 6))
year_summary.plot(
    kind="line",
    marker="o"
)
plt.title("Average Closing Price by Year")
plt.xlabel("Year")
plt.ylabel("Average Closing Price")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
top_10_assets = (
    df.groupby("asset_name")["close"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
)
plt.figure(figsize=(12, 6))
top_10_assets.plot(
    kind="bar"
)
plt.title("Top 10 Assets by Average Closing Price")
plt.xlabel("Asset Name")
plt.ylabel("Average Closing Price")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
volatility_summary = df.groupby("asset_type")["close"].std()
print(volatility_summary)
```

```python
plt.figure(figsize=(10, 6))
volatility_summary.plot(
    kind="bar"
)
plt.title("Closing Price Volatility by Asset Type")
plt.xlabel("Asset Type")
plt.ylabel("Standard Deviation of Closing Price")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.hist(
    df["close"].dropna(),
    bins=40,
    edgecolor="black"
)
plt.title("Distribution of Closing Prices")
plt.xlabel("Closing Price")
plt.ylabel("Number of Records")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
sample_df = df.sample(
    n=min(5000, len(df)),
    random_state=42
)
plt.figure(figsize=(10, 6))
sns.scatterplot(
    x="volume",
    y="close",
    data=sample_df
)
plt.title("Closing Price vs Trading Volume")
plt.xlabel("Trading Volume")
plt.ylabel("Closing Price")
plt.tight_layout()
plt.show()
```

```python
numeric_cols=["open","high","low","close","volume"]
correlation = df[numeric_cols].corr()
```

```python
print(correlation)
```

```python
plt.figure(figsize=(10, 6))
sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("Correlation Heatmap of Financial Variables")
plt.tight_layout()
plt.show()
```
