```python
total_sales=df["Sales"].sum()
total_expenses=df["Expenses"].sum()
total_profit=df["Profit"].sum()

average_sales=df["Sales"].mean()
highest_sales=df["Sales"].max()
```

```python
monthly=df.groupby("Month")[["Sales","Expenses","Profit"]].sum()
```

```python
print("\n========Monthly summary========")
print(monthly)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(12,6))
plt.plot(
    monthly.index,
    monthly["Sales"],
    marker="o",
    alpha=0.7,
    label="Sales"
)

plt.title("distribution of Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.legend()
plt.grid=True
plt.show()
```

```python
plt.figure(figsize=(12,6))
plt.plot(
    monthly.index,
    monthly["Profit"],
    marker="o",
    alpha=0.7,
    label="Profit"
)

plt.title("distribution of Profit")
plt.xlabel("Month")
plt.ylabel("Profit")
plt.xticks(rotation=45)
plt.grid=True
plt.legend()
plt.show()
```
