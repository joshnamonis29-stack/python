```python
df=df.sort_values("date")
df
```

```python
df["MA_5"]=df["debit_amount"].rolling(5).mean()
df["MA_20"]=df["debit_amount"].rolling(20).mean()
df
```

```python
plt.figure(figsize=(8,6))
plt.plot(df["date"],df["debit_amount"],label="debit_amount price",color="red",marker="o")
plt.plot(df["date"],df["MA_5"],label="5-day moving",color="blue")
plt.plot(df["date"],df["MA_20"],label="5-day moving",color="black")

plt.title("moving average and ledger")
plt.xlabel("moving average")
plt.ylabel("ledger")
plt.grid(True)
plt.legend(title="subjet")
plt.xticks(rotation=45)
plt.show()
```
