```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8,6))
plt.plot(df["Year"],df["EPS"],marker="o")
plt.title("EPS trend")
plt.xlabel("Year")
plt.ylabel("EPS")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8,6))
plt.plot(df["Year"],df["Current_Ratio"],marker="o")
plt.title("Current_Ratio trend")
plt.xlabel("Year")
plt.ylabel("Current_Ratio")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8,6))
plt.plot(df["Year"],df["ROE"],marker="o")
plt.title("ROE trend")
plt.xlabel("Year")
plt.ylabel("ROE")
plt.grid(True)
plt.show()
```
