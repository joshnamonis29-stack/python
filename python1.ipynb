```python
cost_price=int(input("Enter Cost Price:"))
selling_price=int(input("Enter Selling Price:"))
if selling_price>cost_price:
    profit=selling_price-cost_price
    profit=(profit/cost_price)*100
    print("profit:",profit,"Profit Percentage:",profit)
elif selling_price<cost_price:
    loss=cost_price-selling_price
    loss=(loss/cost_price)*100
    print("Loss:",loss,"Loss Percentage:",loss)
else:
    print("No Profit,No Loss") 
```

```python

cost_price=float(input("Enter Cost Price: "))
selling_price=float(input("Enter Base Selling Price: "))
discount_rate=float(input("Enter Discount Rate (%): "))
gst_rate=float(input("Enter GST Rate (%): "))
if selling_price > cost_price:
    print(f"Initial State: Profit of ₹{selling_price - cost_price}")
else:
    print(f"Initial State: Loss of ₹{cost_price - selling_price}")
discount_amount = selling_price * discount_rate / 100
price_after_discount = selling_price - discount_amount
print("Price after Discount: ₹", price_after_discount)
gst_amount = price_after_discount * gst_rate / 100
final_price = price_after_discount + gst_amount
print("Final Price after GST: ₹", final_price)   
```
